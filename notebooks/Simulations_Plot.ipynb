```julia
μs_1=readdlm("data/artificial_data/μs_pos_1.txt");
μs_2=readdlm("data/artificial_data/μs_pos_2.txt");
μs_3=readdlm("data/artificial_data/μs_pos_3.txt");
σs_1=readdlm("data/artificial_data/σs_pos_1.txt");
σs_2=readdlm("data/artificial_data/σs_pos_2.txt");
σs_3=readdlm("data/artificial_data/σs_pos_3.txt");
trajectory_pos_1=readdlm("data/artificial_data/trajectory_pos_1.txt")
trajectory_pos_2=readdlm("data/artificial_data/trajectory_pos_2.txt")
trajectory_pos_3=readdlm("data/artificial_data/trajectory_pos_3.txt");
```

```julia
T0=30.0
m=length(μs_1)
p1=Plots.plot(1:length(trajectory_pos_1),trajectory_pos_1,line=3,color=:red,ylabel="",xlabel="",label="")

Plots.plot!([T0],seriestype="vline",linestyle=:dash,color=:black,label=L"t_0")
Plots.plot!( T0:T0+m-1,μs_1, color=:pink,linestyle=:dash, ribbon=σs_1,label="sampled paths")


plot!(p1,tickfontsize=tfs, legendfontsize=lfs, guidefontsize=alfs,titlefontsize=alfs)
plot!(p1,frame=:box,fmt=:svg,ylim=[0,1])
plot!(p1, size=(500,250),yticks=[0,1],xticks=:false,grid=false,fmt=:svg,)
```

```julia
T0=30.0
m=length(μs_2)
p2=Plots.plot(1:length(trajectory_pos_2),trajectory_pos_2,line=3,color=:gray,ylabel="",xlabel="",label="")

Plots.plot!([T0],seriestype="vline",linestyle=:dash,color=:black,label=L"t_0")
Plots.plot!( T0:T0+m-1,μs_2,  color=:gray,linestyle=:dash, ribbon=σs_2,label="sampled paths")


plot!(p2,tickfontsize=tfs, legendfontsize=lfs, guidefontsize=alfs,titlefontsize=alfs)
plot!(p2,frame=:box,fmt=:svg,ylim=[0,1])
plot!(p2, size=(500,250),yticks=[0,1],xticks=:false,grid=false,fmt=:svg,)
```

```julia
T0=30.0
m=length(μs_3)
p3=Plots.plot(1:length(trajectory_pos_3),trajectory_pos_3,line=3,color=:blue,ylabel="",label="")

Plots.plot!([T0],seriestype="vline",linestyle=:dash,color=:black,label=L"t_0")
Plots.plot!( T0:T0+m-1,μs_3,  color=:blue,linestyle=:dash, ribbon=σs_3,label="sampled paths")


plot!(p3,xlabel="Time (serial interval)",tickfontsize=tfs, legendfontsize=lfs, guidefontsize=alfs,titlefontsize=alfs)
plot!(p3,frame=:box,fmt=:svg,ylim=[0,1])
plot!(p3, size=(500,250),yticks=[0,1],xticks=:true,grid=false,fmt=:svg,)
```

```julia
benef=readdlm("data/artificial_data/Beneficial_inf.txt");
neutral=readdlm("data/artificial_data/Neutral_inf.txt");
delet=readdlm("data/artificial_data/Deleterious_inf.txt");
```

```julia
rang_x=[-0.1,-0.05,0,0.05,0.1];
my_rang_x=rang_x*100;
rang_y=[0.0,0.05,0.1,0.15]
my_rang_y=rang_y*100;
```

```julia

w=ones(1000)*1/1000
p=histogram(benef,bins=100,label="Beneficial",alpha=0.3,weights=w,color=:red)
histogram!(neutral,bins=100,label="Neutral",alpha=0.3,weights=w,color=:gray)
histogram!(delet,bins=100,label="Deleterious",alpha=0.3,color=:blue,weights=w)
Plots.plot!([0.03],seriestype="vline",linestyle=:dash,color=:red,alpha=0.6,line=3,label="")
Plots.plot!([0.0],seriestype="vline",linestyle=:dash,color=:gray,alpha=0.6,line=3,label="")
Plots.plot!([-0.03],seriestype="vline",linestyle=:dash,color=:blue,alpha=0.6,line=3,label="")
plot!(p,xticks=(rang_x,my_rang_x))
plot!(p,yticks=(rang_y,my_rang_y))


plot!(p,tickfontsize=tfs, legendfontsize=lfs, guidefontsize=alfs,titlefontsize=alfs)
plot!(p,frame=:box,fmt=:svg)
plot!(p, size=(600,400),grid=true,fmt=:svg)
#plot!(p,legend_marker=:circle)

ylabel!(latexstring("\nFrequency (%)"))
xlabel!(latexstring("\ninferred  selection coefficients, ŝ (%)\n "))
```
