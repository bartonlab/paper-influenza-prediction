```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
import plotting_functions
```

```python
freq_vs_time = np.loadtxt(
    "data_Figs/Fig_frequency_Toy_model/freq_vs_time_toy_model_example_q_4_L_20_T_1000_mu_1e-3.txt"
)
```

```python
T=1000.0
q=4
L=20

# Create figure
#fig, ax = plt.subplots(figsize=(6, 2))  # similar to (800, 200)
fig, ax = plt.subplots(figsize=plotting_functions.get_figsize("horizontal"))  # single-column width, 2 inch height
#fig, ax = plt.subplots(figsize=(1.5*plotwidth, 0.5*plotwidth))

# Plot all trajectories
for i in range(1, L + 1):
    for s in range(1, q + 1):
        traj = plotting_functions.select_traj(freq_vs_time, i, s, q)
        ax.plot(
            range(1, len(traj) + 1),
            traj,
            lw=1,
            label=""  # No legend
        )

# Customize axes
ax.set_xlabel("Date (Generations)")
ax.set_ylabel("Allele frequency")
ax.set_ylim(0, 1.0)
ax.set_xlim(-10.0, T+10)  # where T is the last time point
#ax.set_xlim(left=-10.0) 
#ax.set_ylim(right=1000.0) 

ax.set_yticks([0.0, 1.0])
ax.set_yticks([0.25, 0.5, 0.75], minor=True)
ax.tick_params(axis="both")
ax.grid(alpha=0.0)

# Remove plot frame
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

# Adjust layout
fig.subplots_adjust(bottom=0.2, left=0.1)
plt.tight_layout()

# Save or show
plt.savefig("Allele_frequency_simulated_data.pdf", bbox_inches='tight',dpi=300)
plt.show()
```
