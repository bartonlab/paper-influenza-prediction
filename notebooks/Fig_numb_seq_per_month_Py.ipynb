```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib
from pathlib import Path
from matplotlib import rcParams
import plotting_functions
```

```python
# ========== Load data ==========
base_path = "data_Figs/Fig_Number_of_sequences/"

num_seqs_H3N2_HA = np.loadtxt(base_path + "H3N2/num_seqs_H3N2_HA.txt")
num_seqs_H3N2_NA = np.loadtxt(base_path + "H3N2/num_seqs_H3N2_NA.txt")
date_labels_H3N2 = np.loadtxt(base_path + "H3N2/date_labels_H3N2.txt", dtype=str)

num_seqs_H1N1_HA = np.loadtxt(base_path + "H1N1/num_seqs_H1N1_HA.txt")
num_seqs_H1N1_NA = np.loadtxt(base_path + "H1N1/num_seqs_H1N1_NA.txt")
date_labels_H1N1 = np.loadtxt(base_path + "H1N1/date_labels_H1N1.txt", dtype=str)
```

```python
# ========== Plot parameters ==========
points_interval = 14
fig, axs = plt.subplots(2, 1, figsize=plotting_functions.get_figsize("square"), dpi=300, constrained_layout=True)

# Colors and markers
ha_color = "#d62728"   # dark red
na_color = "#2ca02c"   # dark green
ha_marker = 'o'
na_marker = '^'

# Plot H3N2
axs[0].plot(num_seqs_H3N2_HA, marker=ha_marker, markersize=2, alpha=0.5,
            linewidth=1, color=ha_color, label='HA')
axs[0].plot(num_seqs_H3N2_NA, marker=na_marker, markersize=2, alpha=0.5,
            linewidth=1, color=na_color, label='NA')
axs[0].set_yscale('log')
axs[0].set_ylabel("Number of sequences\nper month")
axs[0].set_xticks(range(0, len(date_labels_H3N2), points_interval))
axs[0].set_xticklabels(date_labels_H3N2[::points_interval], rotation=35)
axs[0].tick_params(axis='y',)
axs[0].legend(loc=(0.8, 0.1), frameon=False)
axs[0].spines['top'].set_visible(False)
axs[0].spines['right'].set_visible(False)
axs[0].spines['left'].set_visible(True)
axs[0].spines['bottom'].set_visible(True)
axs[0].set_ylim(0.5, 50000)
axs[0].grid(False)
axs[0].set_title("a. H3N2", loc='left')

# Plot H1N1
axs[1].plot(num_seqs_H1N1_HA, marker=ha_marker, markersize=2, alpha=0.5,
            linewidth=1, color=ha_color, label='HA')
axs[1].plot(num_seqs_H1N1_NA, marker=na_marker, markersize=2, alpha=0.5,
            linewidth=1, color=na_color, label='NA')
axs[1].set_yscale('log')
axs[1].set_ylabel("Number of sequences\nper month")
axs[1].set_xlabel("Date (months)")
axs[1].set_xticks(range(0, len(date_labels_H1N1), points_interval))
axs[1].set_xticklabels(date_labels_H1N1[::points_interval], rotation=35)
axs[1].tick_params(axis='y')
axs[1].legend(loc=(0.8, 0.1), frameon=False)
axs[1].spines['top'].set_visible(False)
axs[1].spines['right'].set_visible(False)
axs[1].spines['left'].set_visible(True)
axs[1].spines['bottom'].set_visible(True)
axs[1].set_ylim(0.5, 50000)
axs[1].grid(False)
axs[1].set_title("b. H1N1", loc='left')

# Optional: Save figure
#plt.savefig("Number_of_sequences_per_month_both_subtypes.pdf")

plt.show()
```
