```python
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
import os
from matplotlib import rcParams
import plotting_functions
```

```python

# === Settings ===
base = 9
matplotlib.rcParams.update({
    'font.size': base,
    'axes.labelsize': base,
    'axes.titlesize': base,
    'xtick.labelsize': base,
    'ytick.labelsize': base,
    'legend.fontsize': base,
    'font.family': 'Arial',
})
ms = 6
lw = 2
```

```python
# Define constants
T0 = 300
Tfs = np.arange(90, 361, 30)
data_dir = "data_Figs/Fig_EMD_vs_horizon_Toy_model/emd_sim/"
# Load all data
sim_dists  = [plotting_functions.load_emd_data("medoid_dist", T0, Tf,data_dir) for Tf in Tfs]
null_dists = [plotting_functions.load_emd_data("null_emd", T0, Tf,data_dir) for Tf in Tfs]
lb_dists   = [plotting_functions.load_emd_data("lb_emd", T0, Tf,data_dir) for Tf in Tfs]

# Plot
fig, ax = plt.subplots(figsize=(6,3))

ax.plot(Tfs, null_dists, label="null", lw=2, marker='o', color='gray', alpha=1.0)
ax.plot(Tfs, sim_dists, label="prediction", lw=2, marker='D', color=(0.3, 0.45, 0.8), alpha=1.0)
ax.plot(Tfs, lb_dists, label="lower bound", lw=2, marker='^', color='green', alpha=1.0)

ax.set_xlabel("Forecast horizon (model generations)", fontsize=9)
ax.set_ylabel("Distance to future (symbols)", fontsize=9)
ax.set_xticks(Tfs)
ax.set_ylim(0, 6)

ax.legend(frameon=False)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_visible(True)
ax.spines['bottom'].set_visible(True)

plt.tight_layout()
plt.savefig("EMD_simulated_data_py.pdf")
```
