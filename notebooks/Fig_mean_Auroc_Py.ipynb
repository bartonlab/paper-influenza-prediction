```python
import numpy as np
import pandas as pd
import os
import matplotlib.pyplot as plt
import matplotlib
from pathlib import Path
from matplotlib import rcParams
import plotting_functions
```

```python
# Load data
null_list = np.loadtxt("data_Figs/Fig_AUC_by_methods/null_list_full.txt")
sim_list = np.loadtxt("data_Figs/Fig_AUC_by_methods/sim_list_full.txt")
sim_list_SLM = np.loadtxt("data_Figs/Fig_AUC_by_methods/sim_list_SLM_IS.txt")
sim_list_SLFS = np.loadtxt("data_Figs/Fig_AUC_by_methods/sim_list_SLFS.txt")
sim_list_FMSS = np.loadtxt("data_Figs/Fig_AUC_by_methods/sim_list_FMSS.txt")
```

```python
variants = ["H3N2-HA", "H3N2-NA", "H1N1-HA", "H1N1-NA"]
lw=2
x = np.arange(len(variants))  # [0, 1, 2, 3]

# Initialize plot
fig, ax = plt.subplots(figsize=plotting_functions.get_figsize("vertical"),dpi=300)  # size in inches

# Plot each series
ax.plot(x, null_list, marker='s', color='gray', linewidth=lw, alpha=1.0, label="Null")
ax.plot(x, sim_list, marker='o', color=(0.3, 0.45, 0.8), linewidth=lw, alpha=1.0, label=" FM")
ax.plot(x, sim_list_SLM, marker='h', color='purple', linewidth=lw, alpha=1.0, label=" SLM")
ax.plot(x, sim_list_FMSS, marker='D', color='black', linewidth=lw, alpha=1.0, label="FM+SS")
ax.plot(x, sim_list_SLFS, marker='^', color='brown', linewidth=lw, alpha=1.0, label="SLM+FS")

# Formatting
ax.set_xticks(x)
ax.set_xticklabels(variants)
ax.set_yticks(np.arange(0.5, 1.01, 0.25))
ax.set_ylim(0.4, 1.0)
ax.set_xlabel("Variant type")
ax.set_ylabel(r"$\langle \mathrm{AUROC} \rangle$")  # LaTeX-style label
ax.legend(loc='upper left', frameon=False)
ax.grid(False)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Tight layout with small margin
#plt.subplots_adjust(left=0.1, right=0.97, top=0.95, bottom=0.2)
plt.savefig("av_AUC_all_proteins_Py.pdf", dpi=300, bbox_inches='tight')
# Show or save the plot
plt.show()
```
