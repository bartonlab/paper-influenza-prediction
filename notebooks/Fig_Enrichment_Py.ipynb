```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib
from pathlib import Path
from matplotlib import rcParams
import plotting_functions
import json
```

```python
#Load Data
with open("data_Figs/Fig_Enrichment/results/H3N2/HA/enrichment_dict.json") as f:
    enrichment_dict_H3N2_HA = json.load(f)
with open("data_Figs/Fig_Enrichment/results/H3N2/NA/enrichment_dict.json") as f:
    enrichment_dict_H3N2_NA = json.load(f)
with open("data_Figs/Fig_Enrichment/results/H1N1/HA/enrichment_dict.json") as f:
    enrichment_dict_H1N1_HA = json.load(f)   
with open("data_Figs/Fig_Enrichment/results/H1N1/NA/enrichment_dict.json") as f:
    enrichment_dict_H1N1_NA = json.load(f)
# Now enrichment_dict is a Dict[str, List[float]]
```

```python
fig, axs = plt.subplots(
    2, 2,
    figsize=plotting_functions.get_figsize("vertical"),
    constrained_layout=False
)

# Plotting code
plotting_functions.get_enrichment_plot_H3N2_HA(enrichment_dict_H3N2_HA, axs[0, 0], show_xlabel=False, show_ylabel=False)
plotting_functions.get_enrichment_plot_H1N1_HA(enrichment_dict_H1N1_HA, axs[0, 1], show_xlabel=False, show_ylabel=False)
plotting_functions.get_enrichment_plot_H3N2_NA(enrichment_dict_H3N2_NA, axs[1, 0], show_xlabel=False, show_ylabel=False)
plotting_functions.get_enrichment_plot_H1N1_NA(enrichment_dict_H1N1_NA, axs[1, 1], show_xlabel=False, show_ylabel=False)

# Adjust spacing
plt.subplots_adjust(left=0.1, right=0.95, top=0.95, bottom=0.15, wspace=0.3, hspace=0.4)

# Add shared axis labels
fig.text(0.5, 0.04, 'Fraction of most positively selected variants (%)', ha='center', va='center')
fig.text(0.04, 0.5, 'Fold enrichment for epitope positions', ha='center', va='center', rotation='vertical')
plt.savefig("Enrichment_all_proteins_t2009_py.pdf")

plt.show()
```
